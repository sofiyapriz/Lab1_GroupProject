```python
# Побудуйте графік автокореляційної функції (ACF)
plot_acf(df["temp"], lags=60)
plt.show()
```

Автокореляційна функція (ACF) використовується для аналізу залежності значень часового ряду від його попередніх значень (лагів).

На графіку ACF по горизонтальній осі відкладаються лаги (у моєму випадку до 60 кроків назад), а по вертикальній значення коефіцієнта автокореляції.

Це дозволяє:
*  виявити наявність тренду, сезонності або циклічності;
* оцінити глибину часової залежності показника;
* визначити параметри статистичних моделей прогнозування, зокрема порядку MA (моделі ковзного середнього) у межах ARIMA.

```python
# Побудуйте графік часткової автокореляційної функції (PACF)
plot_pacf(df["temp"], lags=60)
plt.show()
```

Часткова автокореляційна функція (PACF) застосовується для вимірювання «чистої» залежності між значеннями часового ряду.

На відміну від звичайної ACF, PACF ізолює зв'язок між поточним спостереженням і значенням з певним лагом, усуваючи опосередкований вплив усіх проміжних лагів.

На графіку PACF (для обраних lags=60) по горизонтальній осі відкладаються часові зсуви, а по вертикальній коефіцієнти часткової кореляції.


Це дозволяє:
* визначити лаги, які мають безпосередній статистично значущий вплив на поточне значення;
* встановити порядок авторегресійної складової (p) для побудови моделей прогнозування, зокрема моделі ARIMA.

```python
# Сформуйте цільовий часовий ряд та розділіть його на train/test у часовому порядку
target = df["temp"].copy()

split_index = int(len(target) * 0.8)
train = target.iloc[:split_index]
test = target.iloc[split_index:]

print("Train size:", len(train))
print("Test size:", len(test))
```

ARIMA

https://developer.ibm.com/tutorials/awb-arima-models-in-python/

```python
# Побудуйте та навчіть модель ARIMA. Отримайте прогноз для тестової вибірки
train_arima = train.reset_index(drop=True)

arima_model = ARIMA(train_arima, order=(5, 1, 2))
arima_fit = arima_model.fit()

arima_forecast = arima_fit.forecast (steps=len(test))
arima_forecast.index = test.index

print(arima_forecast.head())
```

```python
# Виведіть статистичний звіт навченої ARIMA-моделі
print(arima_fit.summary())
```

сигма 2 : це представляє дисперсію залишкових значень, чим менше, тим краще

логарифмічна правдоподібність : це оцінка максимальної правдоподібності для моделі ARIMA, чим вище, тим краще

aic : міра ступеню відповідності та економності моделі, чим нижче, тим краще

```python
# Проаналізуйте залишки ARIMA-моделі та побудуйте відповідні графіки
#density plot of residuals

import pandas as pd

residuals = pd.DataFrame(arima_fit.resid)
fig, ax = plt.subplots(1, 2, figsize=(14, 6))
residuals.plot(title="Residuals", ax=ax[0])
residuals.plot(kind='kde', title='Density', ax=ax[1])

plt.show()
```

```python
# Реалізуйте MAPE та обчисліть MAE, RMSE, MAPE для ARIMA
def mape(y_true, y_pred):
  y_true = np.array(y_true)
  y_pred = np.array(y_pred)
  mask = y_true != 0
  return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100

arima_mae = mean_absolute_error(test, arima_forecast)
arima_rmse = np.sqrt(mean_squared_error(test, arima_forecast))
arima_mape = mape(test, arima_forecast)

print("ARIMA MAE:", arima_mae)
print("ARIMA RMSE:", arima_rmse)
print("ARIMA MAPE:", arima_mape)
```

```python
import matplotlib.pyplot as plt

# Застосовуємо сучасний чистий стиль
plt.style.use('seaborn-v0_8-whitegrid')

fig, ax = plt.subplots(figsize=(14, 6))

ax.plot(train.index, train, label="Навчальна вибірка (Train)", color="#2ca02c", linewidth=1.5)
ax.plot(test.index, test, label="Тестова вибірка (Test)", color="#1f77b4", linewidth=1.5)
ax.plot(arima_forecast.index, arima_forecast, label="Прогноз ARIMA", color="#FF0000", linestyle="--", linewidth=2)

ax.set_title("Прогноз температури моделі ARIMA", fontsize=14, fontweight="bold", pad=15)
ax.set_xlabel("Дата", fontsize=12)
ax.set_ylabel("Температура", fontsize=12)

ax.legend(frameon=True, facecolor="white", edgecolor="none", fontsize=11)
ax.grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```

PROPHET

https://facebook.github.io/prophet/docs/quick_start.html

```python
# Підготуйте дані у форматі, необхідному для Prophet
prophet_df = df[['temp']].reset_index()
prophet_df.columns = ['ds', 'y']
prophet_train = prophet_df.iloc[:split_index]
prophet_test = prophet_df.iloc[split_index:]
```

```python
# Створіть та навчіть модель Prophet
from prophet import Prophet


prophet_model = Prophet (
    daily_seasonality=False,
    weekly_seasonality=True,
    yearly_seasonality=True
)

prophet_model.fit(prophet_train)
```

Прогноз

```python
# Побудуйте прогноз Prophet для тестового періоду
future = prophet_model.make_future_dataframe (periods=len (prophet_test), freq="D")
prophet_forecast_full = prophet_model.predict(future)
prophet_forecast = prophet_forecast_full[["ds", "yhat"]].tail(len(prophet_test)).copy()
prophet_forecast = prophet_forecast.set_index("ds")
prophet_test_indexed  = prophet_test.set_index("ds")
```

```python
# Обчисліть MAE, RMSE, MAPE для Prophet
prophet_mae = mean_absolute_error(prophet_test_indexed["y"], prophet_forecast["yhat"])
prophet_rmse = np.sqrt(mean_squared_error(prophet_test_indexed["y"], prophet_forecast["yhat"]))
prophet_mape = mape(prophet_test_indexed["y"], prophet_forecast["yhat"])

print("Prophet MAE:", prophet_mae)
print("Prophet RMSE:", prophet_rmse)
print("Prophet MAPE:", prophet_mape)
```

```python
# Візуалізуйте фактичні значення та прогноз Prophet
plt.figure(figsize=(14, 6))

# Тренувальні дані
plt.plot(prophet_train["ds"], prophet_train["y"], label="Train", color="#2ca02c", linewidth=1.5)

# Тестові дані
plt.plot(prophet_test["ds"], prophet_test["y"], label="Test", color="#1f77b4", linewidth=1.5)

# Прогноз
plt.plot(prophet_forecast.index, prophet_forecast["yhat"], label="Prophet Forecast", color="#FF0000", linewidth=2)


plt.title("Прогноз Prophet", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Дата", fontsize=12)
plt.ylabel("Температура", fontsize=12)
plt.legend(frameon=True, facecolor="white", edgecolor="none", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.6)

# Показуємо графік
plt.tight_layout()
plt.show()
```

```python
# Візуалізуйте фактичні значення та прогноз
# Вбудований метод Prophet
fig = prophet_model.plot(prophet_forecast_full)   # де prophet_model - модель Prophet, а prophet_forecast_full - результат прогнозів
plt.title("Прогноз Prophet")
plt.show()
```

ПІДГОТОВКА ДАНИХ ДЛЯ LSTM

```python
# Масштабуйте часовий ряд та сформуйте послідовності для LSTM
scaler = MinMaxScaler()
scaled_values = scaler.fit_transform(df[["temp"]])

sequence_length = 30
X = []
y= []

for i in range(sequence_length, len(scaled_values)):
  X.append(scaled_values[i-sequence_length:i, 0])
  y.append(scaled_values[i, 0])


X = np.array(X)
y= np.array(y)
X = X.reshape((X.shape[0], X.shape[1], 1))

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Розділіть послідовності на тренувальну та тестову вибірки без перемішування
split_lstm = int(len(X) * 0.8)

X_train = X[:split_lstm]
X_test = X[split_lstm:]
y_train = y[:split_lstm]
y_test = y[split_lstm:]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
# Побудуйте архітектуру LSTM-моделі та скомпілюйте її
lstm_model = Sequential()
lstm_model.add(LSTM (50, return_sequences=True, input_shape=(X_train.shape[1], 1)))
lstm_model.add(LSTM(50))
lstm_model.add(Dense(1))
lstm_model.compile(optimizer="adam", loss="mse")
```

```python
# Навчіть LSTM-модель
history = lstm_model.fit(
    X_train,
    y_train,
    epochs=15,
    batch_size=32,
    validation_data=(X_test, y_test),
    verbose=1
)
```

```python
# Виведіть структуру LSTM-моделі
lstm_model.summary()
```

```python
# Отримайте прогноз LSTM та виконайте зворотне масштабування
lstm_pred_scaled = lstm_model.predict(X_test)
lstm_pred = scaler.inverse_transform(lstm_pred_scaled)
y_test_real = scaler.inverse_transform(y_test.reshape(-1, 1))
```

```python
# Сформуйте часові мітки для тестового прогнозу LSTM
lstm_dates = df.index[sequence_length:][split_lstm:]
```

```python
# Обчисліть MAE, RMSE, MAPE для LSTM
lstm_mae = mean_absolute_error(y_test_real, lstm_pred)
lstm_rmse = np.sqrt(mean_squared_error(y_test_real, lstm_pred))
lstm_mape = mape(y_test_real.flatten(), lstm_pred.flatten())

print("LSTM MAE:", lstm_mae)
print("LSTM RMSE:", lstm_rmse)
print("LSTM MAPE:", lstm_mape)
```

```python
# Візуалізуйте фактичні значення та прогноз LSTM
plt.figure(figsize=(14, 6))
plt.plot(lstm_dates, y_test_real, label="Actual")
plt.plot(lstm_dates, lstm_pred, label="LSTM Forecast")

plt.title("Прогноз LSTM")
plt.xlabel("Дата")
plt.ylabel("Температура")
plt.legend()
plt.grid(True)
plt.show()
```

Порівняння моделей

```python
# Створіть підсумкову таблицю MAE, RMSE, MAPE для ARIMA, Prophet та LSTM
# Порівняйте моделі за отриманими результатами
results = pd.DataFrame({
    "Model": ["ARIMA", "Prophet", "LSTM"],
    "MAE": [arima_mae, prophet_mae, lstm_mae],
    "RMSE": [arima_rmse, prophet_rmse, lstm_rmse],
    "MAPE": [arima_mape, prophet_mape, lstm_mape]
}).set_index("Model").sort_values("RMSE")

print(results)
```
